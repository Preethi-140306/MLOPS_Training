```python
import pandas as pd
df = pd.read_csv('cleaned_women_safety.csv')
df.head()
```

```python
dummies=pd.get_dummies(df.label)
dummies
```

```python
merged = pd.concat([df,dummies],axis='columns')
merged
```

```python
final = merged.drop(['label'],axis='columns')
final
```

```python
final=final.drop([2], axis='columns')
final
```

```python
X=final.drop(['text'],axis='columns')
X
```

```python
y=df.label
```

```python
from sklearn.linear_model import LinearRegression

model = LinearRegression()
model.fit(X, y)
print(model.predict(X))
print(model.score(X,y))
```

```python
from sklearn.preprocessing import LabelEncoder

le = LabelEncoder()
dfle = df.copy()

dfle.label = le.fit_transform(dfle.label)
dfle
```

```python
X = dfle[['label']].values
X
```

```python
y = dfle.label.values
y
```

```python
from sklearn.preprocessing import OneHotEncoder
from sklearn.compose import ColumnTransformer

ct = ColumnTransformer(
    [('label', OneHotEncoder(sparse_output=False), [0])],
    remainder='passthrough'
)
X = ct.fit_transform(X)
X
```

```python
X = X[:, 1:]
X
```

```python
from sklearn.linear_model import LinearRegression
model = LinearRegression()
model.fit(X, y)
model.predict(X)
```

```python
import joblib
joblib.dump(ct, 'encoder.joblib')
joblib.dump(model, 'model.joblib')
model = joblib.load('model.joblib')
ct = joblib.load('encoder.joblib')
print("The OneHotEncoder is saved as joblib file successfully")
```
