```python
import pandas as pd
df=pd.read_csv('cleaned_women_safety.csv')
df.head()
```

```python
from sklearn.preprocessing import LabelEncoder
le = LabelEncoder()
df['text_encoded'] = le.fit_transform(df['text'])
```

```python
df['text_length'] = df['text'].apply(len)
df['word_count'] = df['text'].apply(lambda x: len(x.split()))
```

```python
X = df[['text_encoded', 'text_length', 'word_count']]
y = df['label']
```

```python
from sklearn.linear_model import LinearRegression
model = LinearRegression()
model.fit(X, y)
y_pred = model.predict(X)
print("Predictions:", y_pred[:10])
```

```python
print("Coefficients:", model.coef_)
print("Intercept:", model.intercept_)
```

```python
import pickle
with open('multiple_linear_regression_model.pkl', 'wb') as file:
    pickle.dump(model, file)

print("Model saved successfully as multiple_linear_regression_model.pkl")
```
