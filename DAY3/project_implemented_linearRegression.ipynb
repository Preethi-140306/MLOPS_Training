```python
import pandas as pd
df = pd.read_csv('cleaned_women_safety.csv')
df.head()
```

```python
import matplotlib.pyplot as plt
from sklearn.preprocessing import LabelEncoder
le = LabelEncoder()
df['text_encoded']=le.fit_transform(df['text'])
plt.scatter(df['text_encoded'],df['label'],color='green',marker='*')
plt.xlabel('text (encoded)')
plt.ylabel('label')
plt.show()
```

```python
import numpy as np
X=df[['text_encoded']]
y=df['label']
from sklearn.linear_model import LinearRegression
model=LinearRegression()
model.fit(X, y)
y_pred=model.predict(X)
y_pred_class=np.round(y_pred)
df['predicted_value']=y_pred_class
df.head(10)
```

```python
import matplotlib.pyplot as plt
plt.scatter(X,y,color='green',label='Actual')
plt.plot(X,y_pred_class,color='blue',label='Predicted')
plt.legend()
plt.show()
```

```python
import pickle
with open('simple_linear_regression_model.pkl', 'wb') as file:
    pickle.dump(model, file)

print("Simple Linear Regression model saved successfully.")
```
