```python
import pandas as pd
import matplotlib.pyplot as plt
from sklearn.datasets import load_iris
from sklearn.cluster import KMeans
from sklearn.preprocessing import StandardScaler
```

```python
iris = load_iris()
```

```python
df = pd.DataFrame(iris.data, columns=iris.feature_names)
df = df[['petal length (cm)', 'petal width (cm)']]
```

```python
inertia_without_scaling = []
for k in range(1,11):
  km = KMeans(n_clusters=k, random_state=42)
  km.fit(df)
  inertia_without_scaling.append(km.inertia_)
```

```python
scaler = StandardScaler()
df_scaled = scaler.fit_transform(df)
```

```python
inertia_with_scaling = []
for k in range(1,11):
  km = KMeans(n_clusters=k, random_state=42)
  km.fit(df_scaled)
  inertia_with_scaling.append(km.inertia_)-
```

```python
plt.figure()
plt.plot(range(1,11),inertia_without_scaling,marker='o',label='without Scaling')
plt.plot(range(1,11),inertia_with_scaling,marker='o',label='With Scaling')
plt.xlabel('Number of Clusters (K)')
plt.ylabel('Inertia (WCSS)')
plt.title('Elbow Method for Optimal K')
plt.legend()
plt.show
```
