```python
import pandas as pd
df = pd.read_csv('cleaned_women_safety.csv')
df.head()
```

```python
import matplotlib.pyplot as plt
from sklearn.preprocessing import LabelEncoder
le = LabelEncoder()
df['text_encoded'] = le.fit_transform(df['text'])
plt.scatter(df.text_encoded,df.label,color='red',marker='*')
plt.show()
```

```python
from sklearn.model_selection import train_test_split
X_train,X_test,y_train,y_test = train_test_split(df[['text_encoded']],df['label'],train_size=0.8)
```

```python
X_test
```

```python
from sklearn.linear_model import LogisticRegression
model=LogisticRegression()
model.fit(X_train,y_train)
```

```python
model.predict(X_test)
```

```python
model.score(X_test,y_test)
```
